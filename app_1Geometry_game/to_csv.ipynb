```python
import pandas as pd
import os
from glob import glob

# Path to the folder containing your CSVs
csv_folder = 'cvs_files'

# Output Excel file path
output_excel = 'combined_output.xlsx'

# Get list of all CSV files
csv_files = glob(os.path.join(csv_folder, '*.csv'))

# Create a Pandas Excel writer
with pd.ExcelWriter(output_excel, engine='openpyxl') as writer:
    for csv_file in csv_files:
        # Use the file name (without extension) as the sheet name
        sheet_name = os.path.splitext(os.path.basename(csv_file))[0][:31]  # Excel sheet name limit
        df = pd.read_csv(csv_file)
        df.to_excel(writer, sheet_name=sheet_name, index=False)

print(f"✅ Combined {len(csv_files)} CSV files into {output_excel}")
```
